```python
#Extract date features

matches['date']=pd.to_datetime(matches['date'])
matches['year']=matches['date'].dt.year
matches['month']=matches['date'].dt.month
matches['day']=matches['date'].dt.day

#drop the original date column
matches.drop('date',axis=1,inplace=True)
```

```python
#split season column. Hun kayi season 2024 hai kayi 2007/08 eve hai. Ehna nu handle karna pena

def split_season(season):
    if '/' in season:
        start,end=season.split('/')
        start=int(start)
        end=int('20' + end) if len(end)==2 else int(end)
    else:
        start=end=int(season)
    return start,end

# aah code ki kar reha- pehle je 2007/08 aa ohnu int of 2007 and 08 ch todd reha the je end di length 2 hai taan ohde agge 20 laa reha
#matlab 2007 2008 bana reha
# te je 2024 hai ohnu 2024 2024 ch todd reha
```

```python
matches[['season_start','season_end']]=matches['season'].apply(lambda x:pd.Series(split_season(x)))
#eh ik value 2008/09 nu chakda te 2 ch convert kar dinda 2008 and 2009 ie season_start and season_end de columns ch 2024 2024 nu 
#season_start 2024 te season_end 2024 de vich change kar dinda
#lambda is just faster way to write a function
```

```python
teams=pd.concat([matches['team1'],matches['team2']]).unique()
team_stats=pd.DataFrame(teams,columns=['team_name'])

team_stats
#pehli line team1 te team2 nu ik kar rahi aa matlab concatinate then oss cho unique kadd rahi aa
#dujji line oh array nu dataframe ch badal rahi aa team_name aale column ch
#array toh dataframe ch iss layi change kitta ke dataframe aale function laa paayie uss te. Array te nahi lagne si
```

```python
#match played and won te win percentage
team_stats['matches_played']=team_stats['team_name'].apply(lambda x:len(matches[(matches['team1']==x) | (matches['team2']==x)]))
team_stats['matches_won']=team_stats['team_name'].apply(lambda x:len(matches[matches['winner']==x]))
team_stats['win_percentage']=(team_stats['matches_won']/team_stats['matches_played'])*100

team_stats
#apply means apply that on every row
#lambda here is simply doing:“Take one team name at a time and calculate something.”
#lambda used jado stand alone function na banaana hove te iko jagah kamm aave
#oh aale method naal krke dekhna bhi pehle =0 then jado jado aave taan +1 kari jaao
```

```python
#hun team wise analysis. total runs scored and total wickets taken by a team

#IMPORTANT- total wickets taken are not actually total wickets taken oh total overs aa.Total wickets taken deliveries aale dataset toh aayega.


team_stats['total_runs_scored'] = team_stats['team_name'].apply(
    lambda x:
        matches[matches['team1'] == x]['target_runs'].sum()
        +
        matches[matches['team2'] == x]['target_runs'].sum()
)


team_stats['total_wickets_taken'] = team_stats['team_name'].apply(
    lambda x:
        matches[matches['team1'] == x]['target_overs'].sum()
        +
        matches[matches['team2'] == x]['target_overs'].sum()
)

team_stats['total_runs_scored']=team_stats['total_runs_scored'].astype(int)
team_stats['total_wickets_taken'] = team_stats['total_wickets_taken'].astype(int)
team_stats
```

```python
#avg_run,economy_rate,run_rate
#oh so it is per delivery in the dataset so no of rows by that team would be equal to the total balls that is why its not mentioned differently somewhere?
#so /(deliveries[deliveries['batting_team']==x].shape[0]/6)) apne aap hi mean no of balls
#deliveries[deliveries['batting_team']=='CSK']means:
              #From deliveries dataframe,
              #keep only rows where batting_team is CSK.

team_stats['average_runs_scored_per_match']=team_stats['total_runs_scored']/team_stats['matches_played']

team_stats['run_rate']=team_stats['team_name'].apply(
    lambda x:
        deliveries[deliveries['batting_team']==x]['total_runs'].sum()/(deliveries[deliveries['batting_team']==x].shape[0]/6))

team_stats['economy_rate']=team_stats['team_name'].apply(
    lambda x:
        deliveries[deliveries['bowling_team']==x]['total_runs'].sum()/(deliveries[deliveries['bowling_team']==x].shape[0]/6))

team_stats
```

```python
#excluding no result matches from deliveries
#matches cho jihna da result nahi si hun oh aale appan deliveries cho drop kar rahe aa
no_result_match_ids=[829813,501265,829813,1359519,1178424]
filtered_deliveries=deliveries[~deliveries['match_id'].isin(no_result_match_ids)]
#last line of code meant ke oh aale kadd do

#highest and lowest runs scored by teams excluding no result matches
#aah groupby function taan laa rahe aa kyonke same match_id aale ch 2 teams batting karengi. Oh 2 team alag alag karke analysis krni aa
#groupby hun filtered deliveries aala le reha and ohnu group kar reha match id te batting team toh
highest_scores=filtered_deliveries.groupby(['match_id','batting_team'])['total_runs'].sum().reset_index()
team_stats['highest_score']=team_stats['team_name'].apply(
    lambda x:
        highest_scores[highest_scores['batting_team']==x]['total_runs'].max())
team_stats['lowest_score']=team_stats['team_name'].apply(
    lambda x:
        highest_scores[highest_scores['batting_team']==x]['total_runs'].min())

team_stats

#shayad its showing victory margin 
```

```python

# Data load aur team mapping (jaise aapne upar kiya tha)
deliveries = pd.read_csv('/Users/shabadpreetsingh/Downloads/ipl data/deliveries.csv')

team_name_mapping = {
    'Delhi Daredevils':'Delhi Capitals',
    'Kings XI Punjab':'Punjab Kings',
    'Rising Pune Supergiants':'Pune Warriors',
    'Rising Pune Supergiant':'Pune Warriors',
    'Gujarat Lions':'Gujarat Titans',
    'Deccan Chargers':'Sunrisers Hyderabad',
    'Royal Challengers Bangalore':'Royal Challengers Bengaluru'
}
deliveries['batting_team'] = deliveries['batting_team'].replace(team_name_mapping)

# No result wale matches nikal diye
no_result_match_ids = [829813, 501265, 1359519, 1178424]
filtered_deliveries = deliveries[~deliveries['match_id'].isin(no_result_match_ids)]

# Har match ki inning ka data nikalna (runs, wickets, aur kitne overs khele)
inning_data = filtered_deliveries.groupby(['match_id', 'inning', 'batting_team']).agg(
    total_runs=('total_runs', 'sum'),
    total_wickets=('is_wicket', 'sum'),
    max_over=('over', 'max') # Yeh check karega ki aakhiri over kaunsa tha
).reset_index()

# HIGHEST SCORES
highest_scores = inning_data.groupby('batting_team')['total_runs'].max().reset_index()
highest_scores.rename(columns={'total_runs': 'Highest_Score'}, inplace=True)

# LOWEST SCORES (Real logic)
# Condition: Ya toh 10 wickets gire hon (| matlab OR) ya fir poore 20 overs khele hon (max_over == 19)
valid_innings = inning_data[(inning_data['total_wickets'] >= 10) | (inning_data['max_over'] == 19)]

lowest_scores = valid_innings.groupby('batting_team')['total_runs'].min().reset_index()
lowest_scores.rename(columns={'total_runs': 'Lowest_Score'}, inplace=True)

# Dono ko merge karna
team_stats = pd.merge(highest_scores, lowest_scores, on='batting_team', how='left')

# Highest score ke hisaab se sort kar diya
team_stats = team_stats.sort_values(by='Highest_Score', ascending=False).reset_index(drop=True)

print("--- Team-wise Highest and Lowest Scores ---")
print(team_stats)
```
